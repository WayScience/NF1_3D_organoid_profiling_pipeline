```python
for patient in tqdm.tqdm(
    data_dict_of_patient_well_fovs,
    desc="Processing patients",
    total=len(data_dict_of_patient_well_fovs),
    leave=True,
):
    for well_fov in tqdm.tqdm(
        data_dict_of_patient_well_fovs[patient],
        desc="Processing wells/fovs",
        total=len(data_dict_of_patient_well_fovs[patient]),
        leave=False,
    ):
        for mask_object in tqdm.tqdm(
            ["nuclei", "cell", "cytoplasm", "organoid"],
            desc="Processing mask objects",
            total=4,
            leave=False,
        ):
            mask_0_1um_path = [
                x
                for x in find_files_available(
                    data_dict_of_patient_well_fovs[patient][well_fov]["0_1"],
                    image_extensions=".tiff",
                )
                if mask_object in pathlib.Path(x).name
            ][0]
            mask_0_1um_path_name = pathlib.Path(mask_0_1um_path).name

            mask_0_2um_path = (
                data_dict_of_patient_well_fovs[patient][well_fov]["0_2"]
                / mask_0_1um_path_name
            )
            mask_0_5um_path = (
                data_dict_of_patient_well_fovs[patient][well_fov]["0_5"]
                / mask_0_1um_path_name
            )
            mask_1um_path = (
                data_dict_of_patient_well_fovs[patient][well_fov]["1"]
                / mask_0_1um_path_name
            )
            for path in [mask_0_2um_path, mask_0_5um_path, mask_1um_path]:
                if not path.parent.exists():
                    path.parent.mkdir(parents=True, exist_ok=True)

            mask_0_1um = read_zstack_image(mask_0_1um_path)

            tifffile.imwrite(mask_0_2um_path, mask_0_1um[::2, :, :].copy())
            tifffile.imwrite(mask_0_5um_path, mask_0_1um[::5, :, :].copy())
            tifffile.imwrite(mask_1um_path, mask_0_1um[::10, :, :].copy())
```
